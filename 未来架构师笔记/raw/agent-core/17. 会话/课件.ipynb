# 最佳实践

在使用会话功能的时候，不管以何种方式实现，应该遵循以下法则：

- **历史上下文不可修改**

  不满足此条，会导致难以命中模型服务商的 prompt cache
- **上下文中应保留思维链**

  不满足此条，会导致推理上下文丢失，从而导致模型回复质量下降
